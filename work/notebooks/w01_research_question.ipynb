## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:**

Can observable search and content signals help us rank which pages should be reviewed for a possible refresh first?

**Decision:**
Which pages should a content team prioritize for review?

**Action:**
A reviewer could inspect the highest-priority pages and decide whether they need actions such as updating outdated information, improving content depth, or investigating search-performance issues.

**Unit of analysis:**
One content page.

**Potential output:**
A ranked list of pages with an opportunity/risk score and reason codes explaining why each page was prioritized.

**Cost of a wrong recommendation:**
If a page is incorrectly prioritized, a reviewer may spend limited time improving a page that does not need attention. More importantly, if an important page is missed, the team may fail to investigate a genuine content or search-performance opportunity.

This is a decision-support problem rather than a claim that a model can guarantee that refreshing a page will improve its performance.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

My work will be able to provide observed and directional insights from the available page-level data. The dataset contains 30,000 pages, with content ages ranging from 90 to 564 days and a median age of 236 days. It also contains 90-day performance signals such as impressions and sessions, which can be used to identify patterns and help prioritize pages for human review.

The analysis may provide decision support by producing a ranked list of pages that appear more worthy of a content-refresh review based on observable signals. For example, the data contains 16,262 pages with a down trend direction, which provides a substantial group that could be investigated.

However, I cannot claim that a page will definitely improve after being refreshed. I cannot establish causality from this observational dataset, and I cannot claim that the analysis is "predicting Google" or predicting Google's ranking decisions. I also cannot claim that a model will generalize to all websites or future data without further validation.

The goal is therefore to investigate whether the available data can provide useful, reliable decision support for prioritizing content-review opportunities, rather than proving that a particular content change will cause better search performance.
