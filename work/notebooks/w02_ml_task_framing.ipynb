## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task type: Classification**

I frame my lane as a classification problem. The goal is to predict whether a page is likely to be underperforming in organic search or not.

The model would assign each page to one of two classes:

- `1` = underperforming page
- `0` = not underperforming page

The prediction can be used as decision support to help prioritize which pages should be reviewed first.

```python
print("Task type: Classification")
print("Goal: Predict whether a page is underperforming or not.")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The ideal target would be whether a page actually needs improvement and whether an intervention would improve its organic search performance. This outcome is not directly observable in the starter data.

Therefore, I use a measurable proxy based on observed CTR. I classify pages with CTR below the dataset median as underperforming and pages at or above the median as not underperforming.

The target is represented as:

1 = underperforming proxy
0 = not underperforming proxy

This is a directional proxy based on observed performance. A lower CTR does not by itself prove that a page needs a specific change because CTR can also be affected by search intent, position, SERP features, and other factors.

```python
import pandas as pd
from pathlib import Path

data_path = Path("flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv")

df = pd.read_csv(data_path)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will use F1 score as the primary success metric.

F1 score balances precision and recall. Precision is important because pages classified as underperforming may be prioritized for review. Recall is important because missing genuinely underperforming pages could cause potentially useful review opportunities to be overlooked.

F1 therefore provides a useful measure for evaluating how well the classifier identifies the underperforming class while balancing false positives and false negatives.

The metric will be used as a measured evaluation metric for decision support, rather than as proof that every predicted page actually needs an intervention.

```python
from sklearn.metrics import f1_score, precision_score, recall_score
print("Primary metric: F1 score")
print("Supporting metrics: Precision and Recall")
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row = one page/content item.

Each row represents an individual page identified by content_id. The row contains observed search, traffic, engagement, freshness, and content characteristics for that page.

The proposed classification target is underperforming, which indicates whether the page falls below the dataset median for CTR.

The model would use page-level signals to classify each page and support prioritization for human SEO/content review.

```python
print("Unit of analysis: one row = one page/content item")
print("Dataset shape:", df.shape)
display(df.head(10))
```

```python
# Create a binary target based on observed CTR performance.
# Pages with CTR below the median are labelled as underperforming.

ctr_threshold = df["ctr"].median()

df["underperforming"] = (df["ctr"] < ctr_threshold).astype(int)

print("CTR threshold:", ctr_threshold)
print("\nClass distribution:")
print(df["underperforming"].value_counts())
```

```python
print("Class percentages:")

class_percentages = (
    df["underperforming"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(class_percentages)
```

```python
display(
    df[
        [
            "content_id",
            "ctr",
            "avg_position",
            "impressions_90d",
            "clicks_90d",
            "engagement_rate",
            "underperforming"
        ]
    ].head(10)
)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could classify pages using CTR alone, such as marking every page below the median CTR as underperforming. This approach is simple and transparent, but it only considers one signal.

ML can combine multiple observed page-level and search-performance signals such as search volume, competition, content type, word count, impressions, clicks, average position, engagement, freshness, and traffic trends.

A classifier could therefore learn patterns across several variables instead of relying on a single manually selected threshold.

The classification output would support a practical content and SEO action: pages predicted as underperforming could be prioritized for human review.

The model would provide decision support rather than automatically deciding what should be changed. Its predictions would be treated as measured and directional signals that help determine which pages deserve further investigation.

```python
print("Decision supported:")
print("Prioritize pages predicted as underperforming for human SEO/content review.")

print("\nWhy use ML:")
print("ML can combine multiple observed signals instead of relying on one fixed rule.")

print("\nImportant limitation:")
print("The classification is decision support and does not prove that a page requires a specific change.")
```
